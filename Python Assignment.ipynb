```python
'''1.	Print all prime numbers between input range (Ex – input 20 50, prints all prime numbers between 20 and 50).'''

n = int(input())
s = int(input())
for i in range(n, s+1):
    if i < 2:
        continue
    for j in range(2, int(i**0.5)+1):
        if i % j == 0:
            break
    else:
        print(i, end = " ")
```

```python
'''2. Factorial using recursion'''

def factorial(s):
    if s == 0 or s == 1:
        return 1
    else:
        return s*factorial(s-1)
s = int(input())
print(factorial(s))
```

```python
'''3.	Square of numbers using lambda'''

s = list(map(int, input().split(',')))
result = list(map(lambda x : x * x, s))
print(result)
```

```python
'''4.	Find the second largest element in a list'''

arr = list(map(int, input().split()))

largest_element = arr[0]
second_element = arr[1]

for i in range(len(arr)):
    if arr[i] > largest_element:
        second_element = largest_element
        largest_element = arr[i]
    elif arr[i] > second_element:
        second_element = arr[i]
print(second_element)
```

```python
'''5.	Count frequency of characters in a string'''

s = input()

n = {}

for i in s:
    if i in n:
        n[i] += 1
    else:
        n[i] = 1
print(n)
```

```python
'''6.	Calculate area of a circle using math library.'''

import math

s = float(input())
area = math.pi*s*s
print(area)
```

```python
'''7.	Reverse a string without using built‑in reverse'''

s = input()
res=""
for i in range(len(s)-1,-1,-1):
    res+=s[i]
print(res)
```

```python
'''8.	Remove duplicates from a list'''

s = input().split(",")

lst = []

for n in s:
    if n not in lst:
        lst.append(n)
print(lst)
```

```python
'''9.	Merge two dictionaries'''

d1 = {'name':'Naveen','Dept':'CSE(CS)'}
d2 = {'name':'Tirupathi','Dept':'AIML'}
for s in d2:
    if s in d1:
        d1[s] = [d1[s], d2[s]]
    else:
        d1[s] = d2[s]
print(d1)
```

```python
'''10.	Fibonacci series using recursion'''

def fibonacci(s):
    if s == 0:
        return 0
    elif s == 1:
        return 1
    else:
        return fibonacci(s-1)+fibonacci(s-2)
s = int(input())
for i in range(s):
    print(fibonacci(i), end = " ")
```
