```python
'''3. Rainwater Collection System
A city installs buildings of different heights along a straight road. 
During rainfall, water gets collected between taller buildings. 
The engineering team needs to calculate the total amount of water 
that can remain trapped after heavy rainfall based on the heights of the buildings.
'''

def water(s):
  left = 0
  right = len(s) - 1
  left_m = 0
  right_m = 0
  ans = 0
  while left < right:
    if s[left] <= s[right]:
      if s[left] >= left_m:
        left_m = s[left]
      else:
        ans += left_m - s[left]
      left += 1
    else:
      if s[right] >= right_m:
        right_m = s[right]
      else:
        ans += right_m - s[right]
      right -= 1
  return ans
s = list(map(int, input("Enter numbers : ").split(",")))
print(water(s))
```

```python
'''12. Mobile Battery Management
A mobile application records the battery consumption and charging values during different time intervals. 
The system needs to identify the longest period during which 
the battery-related values continuously increase according to the recorded data.
'''


def longest(a):
    cur = 1
    ans = 1
    for i in range(1, len(a)):
        if a[i] > a[i-1]:
            cur += 1
        else:
            cur = 1
        ans = max(ans, cur)
    return ans
a = list(map(int, input("Enter values : ").split(",")))
print(longest(a))
```

```python
'''18. Text Editor Correction System
A document editor compares two versions of a sentence. 
Characters can be inserted, removed, or replaced to transform the first version into the second version. 
The system needs to determine the minimum number of such modifications required.
'''


def edit_distance(a, b):
    m = len(a)
    n = len(b)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if a[i-1] == b[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(
                    dp[i][j-1],
                    dp[i-1][j],
                    dp[i-1][j-1]
                )
    return dp[m][n]
a = input("Enter first string : ")
b = input("Enter second string : ")
print(edit_distance(a, b))
```

```python
'''24. Sudoku Puzzle Application
A mobile puzzle application receives a partially completed Sudoku board. 
Empty cells must be filled while ensuring that every row, every column, 
and every 3 × 3 section contains valid numbers without repetition. Develop a solution that completes the puzzle.
'''


def solve(board):
    for r in range(9):
        for c in range(9):
            if board[r][c] == ".":
                for n in "123456789":
                    if valid(board, r, c, n):
                        board[r][c] = n
                        if solve(board):
                            return True
                        board[r][c] = "."
                return False
    return True
def valid(board, r, c, n):
    for i in range(9):
        if board[r][i] == n:
            return False
    for i in range(9):
        if board[i][c] == n:
            return False
    sr = (r // 3) * 3
    sc = (c // 3) * 3
    for i in range(sr, sr + 3):
        for j in range(sc, sc + 3):
            if board[i][j] == n:
                return False
    return True
board = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]

solve(board)
for row in board:
    print(row)
```
